```python
import pandas as pd
from sqlalchemy import create_engine, text

# ── Load all sheets from Excel ──
FILE_PATH = "G:\Amazon_ecommerce.xlsx"  # update path if needed

orders    = pd.read_excel(FILE_PATH, sheet_name="Orders")
customers = pd.read_excel(FILE_PATH, sheet_name="Customers")
products  = pd.read_excel(FILE_PATH, sheet_name="Products")
returns   = pd.read_excel(FILE_PATH, sheet_name="Returns")

# ── Create in-memory SQLite database via SQLAlchemy ──
engine = create_engine("sqlite:///:memory:", echo=False)

orders.to_sql("orders",    engine, index=False, if_exists="replace")
customers.to_sql("customers", engine, index=False, if_exists="replace")
products.to_sql("products",  engine, index=False, if_exists="replace")
returns.to_sql("returns",   engine, index=False, if_exists="replace")

print("✅ All tables loaded successfully!")
print(f"   orders    → {len(orders):,} rows")
print(f"   customers → {len(customers):,} rows")
print(f"   products  → {len(products):,} rows")
print(f"   returns   → {len(returns):,} rows")
```

```python
# Helper function to run SQL and display results
def run_query(sql):
    with engine.connect() as conn:
        df = pd.read_sql(text(sql), conn)
    return df
```

## 📊 Query 1 — Total Revenue, Orders & Avg Order Value by Year

**Business Question:** How has the business grown year over year?

This query tells us the **total revenue earned**, **number of orders placed**, and the **average order value** for each year. It's one of the most important metrics to track business health and growth momentum.

```python
q1 = """
SELECT
    OrderYear                                               AS Year,
    COUNT(OrderID)                                          AS Total_Orders,
    ROUND(SUM(TotalAmount), 2)                              AS Total_Revenue,
    ROUND(AVG(TotalAmount), 2)                              AS Avg_Order_Value,

    ROUND(SUM((UnitPrice * Quantity) - DiscountAmount), 2)  AS Net_Selling_Revenue,

    ROUND(SUM((UnitPrice * Quantity) - DiscountAmount
          - (UnitCost * Quantity)), 2)                      AS Gross_Profit,

    ROUND(
        100.0 * SUM((UnitPrice * Quantity) - DiscountAmount - (UnitCost * Quantity))
        / NULLIF(SUM((UnitPrice * Quantity) - DiscountAmount), 0)
    , 2)                                                     AS Gross_Margin_Pct

FROM orders
GROUP BY OrderYear
ORDER BY OrderYear
"""
run_query(q1)
```

> 💡 **Insight:** The business has literally doubled its revenue in just two years — from ₹1.99 Cr in 2022 to ₹4.10 Cr in 2024. That's a 106% jump which is honestly impressive. But here's what caught my attention — the gross margin has stayed almost flat the whole time (37.16% → 38.19% → 37.49%). Basically the company is selling way more but not actually becoming more efficient. Costs are rising almost as fast as revenue. So yes, top-line growth is great, but unless margins improve, profitability won't grow as fast as the numbers suggest.

## 📦 Query 2 — Top 10 Best-Selling Product Categories by Revenue

**Business Question:** Which product categories are driving the most sales?

Knowing which categories generate the most revenue helps the business decide **where to invest more inventory**, **run targeted promotions**, and **focus marketing spend**. Categories with high revenue but low profit margins may need pricing adjustments.

```python
q2 = """
SELECT
    p.Category,
    COUNT(o.OrderID)               AS Total_Orders,
    SUM(o.Quantity)                AS Units_Sold,
    ROUND(SUM(o.TotalAmount), 2)   AS Total_Revenue,
    ROUND(AVG(o.UnitPrice), 2)     AS Avg_Unit_Price
FROM orders o
JOIN products p ON o.ProductID = p.ProductID
GROUP BY p.Category
ORDER BY Total_Revenue DESC
LIMIT 10
"""
run_query(q2)
```

> 💡 **Insight:** Beauty tops the chart with ₹1.70 Cr in revenue across 341 orders, and Sports is just behind at ₹1.69 Cr. What's interesting is Books actually has the most orders (349) but still earns less than Beauty and Sports — probably because books are cheaper on average. Electronics and Home & Kitchen sit at the bottom with ₹1.33 Cr and ₹1.23 Cr respectively. it might be underpriced or not getting enough visibility.

## 🏆 Query 3 — Top 10 Customers by Total Spend

**Business Question:** Who are our most valuable customers?

The top 10 customers often contribute a disproportionately large share of revenue (Pareto Principle). Identifying them allows the business to offer **VIP treatment, exclusive deals, loyalty rewards**, and dedicated account management to retain them.

```python
q3 = """
SELECT
    c.CustomerID,
    c.CustomerName,
    c.City,
    c.CustomerSegment,
    COUNT(o.OrderID)             AS Total_Orders,
    ROUND(SUM(o.TotalAmount), 2) AS Total_Spend,
    c.LoyaltyPoints
FROM orders o
JOIN customers c ON o.CustomerID = c.CustomerID
GROUP BY c.CustomerID, c.CustomerName, c.City, c.CustomerSegment, c.LoyaltyPoints
ORDER BY Total_Spend DESC
LIMIT 10
"""
run_query(q3)
```

> 💡 **Insight:** The top spender is Deepak Shah from Surat with ₹5.83L across 9 orders. What really stood out to me is that 3 out of the top 10 customers are from Surat — Deepak Shah, Vivek Bose, and Amit Joshi. That's kind of a big deal because it means Surat might be a really strong market worth targeting more. Also Priya Reddy from Mumbai has the most orders (12) and has spent ₹5.25L — she's clearly very loyal. These customers should definitely be put on a VIP plan or given early access to new products.

## 💳 Query 4 — Revenue & Orders by Payment Method

**Business Question:** Which payment methods are customers preferring?

Understanding payment preferences helps the business **optimize checkout flows**, **negotiate better rates with payment partners**, and **identify if certain payment methods have higher order values**. For example, EMI-based purchases often lead to higher basket sizes.

```python
q4 = """
SELECT
    PaymentMethod,
    COUNT(OrderID)                   AS Total_Orders,
    ROUND(SUM(TotalAmount), 2)       AS Total_Revenue,
    ROUND(AVG(TotalAmount), 2)       AS Avg_Order_Value
FROM orders
GROUP BY PaymentMethod
ORDER BY Total_Revenue DESC
"""
run_query(q4)
```

> 💡 **Insight:** Credit Card is the top payment method with ₹1.66 Cr in revenue and the highest average order value of ₹48,872. Cash on Delivery has the most orders (365) but a lower AOV of ₹45,518 — which makes sense since COD customers tend to be more price-conscious. EMI has an AOV of ₹47,051 which proves that when people can pay in installments, they're willing to buy more expensive things. Honestly the platform should push EMI options more at checkout because it clearly drives bigger purchases.

## 📱 Query 5 — Sales Channel Performance (Website vs App vs Other)

**Business Question:** Which sales channel brings in the most revenue?

This shows the **performance of each sales channel** — Mobile App, Website, etc. A growing mobile app channel signals that investing in **app features, push notifications, and app-exclusive offers** can drive significant revenue uplift.

```python
q5 = """
SELECT
    SalesChannel,
    COUNT(OrderID)                   AS Total_Orders,
    ROUND(SUM(TotalAmount), 2)       AS Total_Revenue,
    ROUND(AVG(TotalAmount), 2)       AS Avg_Order_Value,
    ROUND(AVG(DiscountRate) * 100, 2) AS Avg_Discount_Pct
FROM orders
GROUP BY SalesChannel
ORDER BY Total_Revenue DESC
"""
run_query(q5)
```

> 💡 **Insight:** All three channels — Website, Third Party, and Mobile App — are almost neck and neck in revenue (₹3.03 Cr, ₹3.01 Cr, ₹3.00 Cr). But the Mobile App has the highest average order value at ₹46,286, even with the lowest discount rate. This tells me app users are genuinely spending more, not just because of deals. I think investing in the app experience like push notifications and app-only offers would be a smart move to grow that gap.

## 📅 Query 6 — Monthly Revenue Trend (All Years)

**Business Question:** Are there seasonal spikes or dips in sales throughout the year?

Seasonality analysis helps the business **plan inventory ahead of peak months** (e.g., festival seasons, year-end sales), **hire temporary staff**, and **run promotions during slow months** to smoothen revenue. Months like October–December typically see higher e-commerce sales.

```python
q6 = """
SELECT
    OrderYear                        AS Year,
    OrderMonth                       AS Month,
    COUNT(OrderID)                   AS Total_Orders,
    ROUND(SUM(TotalAmount), 2)       AS Monthly_Revenue,
    ROUND(SUM(DiscountAmount), 2)    AS Total_Discounts_Given
FROM orders
GROUP BY OrderYear, OrderMonth
ORDER BY OrderYear, OrderMonth
"""
run_query(q6)
```

> 💡 **Insight:** December 2024 is the biggest single month with ₹47L, and October also does really well every year (₹15.81L in 2022, ₹30.99L in 2023, ₹39.23L in 2024). So there's clearly a festive season effect in Q4. On the other hand, February 2022 is the worst month at only ₹9.52L. What I'd suggest is the business should stock up on inventory 6-8 weeks before October and maybe run clearance sales in February and April to keep revenue steady throughout the year.

## 🔄 Query 7 — Return Rate & Top Return Reasons

**Business Question:** How often are products being returned, and why?

A high return rate eats into profitability. This query helps identify the **most common return reasons** (damaged product, wrong item, late delivery), so the business can take corrective action — better packaging, faster shipping, or stricter quality control. The refund amount shows the **financial impact** of returns.

```python
q7 = """
SELECT
    ReturnReason,
    ReturnStatus,
    COUNT(ReturnID)              AS Total_Returns,
    ROUND(SUM(RefundAmount), 2)  AS Total_Refund_Amount,
    ROUND(AVG(RefundAmount), 2)  AS Avg_Refund_Amount
FROM returns 
GROUP BY ReturnReason,ReturnStatus
ORDER BY Total_Returns DESC
"""
run_query(q7)
```

> 💡 **Insight:** The most common return reason is "Changed Mind" with 33 approved returns and ₹12.59L in refunds, followed by "Wrong Item" with 27 approved returns and ₹9.80L. The "Wrong Item" problem is clearly a dispatch/packaging error — someone is picking the wrong product. "Duplicate Order" has the highest average refund of ₹64,979 per case, which suggests these are big-ticket purchases. Fixing the wrong item dispatch issue alone could save the business close to ₹10L a year in refunds.

## 🌆 Query 8 — Top 10 Cities by Revenue

**Business Question:** Which cities generate the most revenue for us?

Geo-analysis helps the business **prioritize delivery infrastructure, warehousing, and hyperlocal marketing** in high-revenue cities. It also reveals untapped markets in cities with high population but low order counts — potential areas for growth campaigns.

```python
q8 = """
SELECT
    c.City,
    c.State,
    COUNT(o.OrderID)             AS Total_Orders,
    COUNT(DISTINCT o.CustomerID) AS Unique_Customers,
    ROUND(SUM(o.TotalAmount), 2) AS Total_Revenue
FROM orders o
JOIN customers c ON o.CustomerID = c.CustomerID
GROUP BY c.City, c.State
ORDER BY Total_Revenue DESC
LIMIT 10
"""
run_query(q8)
```

> 💡 **Insight:** Kolkata is the top city with ₹1.11 Cr from 228 orders and 63 unique customers — that gives a repeat ratio of about 3.6x per customer, which is the highest in the top 10. Delhi is second at ₹1.01 Cr and Surat third at ₹1.00 Cr. Jaipur is interesting — it has 219 orders (more than Surat's 216) but lower revenue at ₹93.98L. That means Jaipur customers are ordering more frequently but spending less per order. A good upsell campaign targeting Jaipur could actually move the needle a lot.

## 👥 Query 9 — Revenue by Customer Segment

**Business Question:** Which customer segments are most profitable?

Customer segments (Corporate, Home Office, Consumer, etc.) have different buying behaviors and price sensitivities. This analysis guides **segment-specific pricing strategies, bulk discount policies, and B2B vs B2C marketing budgets**.

```python
q9 = """
SELECT
    c.CustomerSegment,
    COUNT(DISTINCT o.CustomerID)    AS Unique_Customers,
    COUNT(o.OrderID)                AS Total_Orders,
    ROUND(SUM(o.TotalAmount), 2)    AS Total_Revenue,
    ROUND(AVG(o.TotalAmount), 2)    AS Avg_Order_Value,
    ROUND(AVG(o.DiscountRate)*100, 2) AS Avg_Discount_Pct
FROM orders o
JOIN customers c ON o.CustomerID = c.CustomerID
GROUP BY c.CustomerSegment
ORDER BY Total_Revenue DESC
"""
run_query(q9)
```

## 💰 Query 11 — Profit Margin Analysis by Product Category

**Business Question:** Which categories are most and least profitable?

Revenue alone is not enough — a category could have high revenue but thin margins. This query shows **gross profit and margin percentage** per category, helping the business decide where to **cut costs, renegotiate supplier prices**, or **increase selling prices**.

```python
q11 = """
SELECT
    p.Category,
    ROUND(SUM((o.UnitPrice * o.Quantity) - o.DiscountAmount), 2)
                                                           AS Net_Selling_Revenue,
    ROUND(SUM(o.UnitCost * o.Quantity), 2)               AS Total_COGS,
    ROUND(SUM((o.UnitPrice * o.Quantity) - o.DiscountAmount
          - (o.UnitCost * o.Quantity)), 2)                  AS Gross_Profit,
    ROUND(
        100.0 * SUM((o.UnitPrice * o.Quantity) - o.DiscountAmount
                       - (o.UnitCost * o.Quantity))
        / NULLIF(SUM((o.UnitPrice * o.Quantity) - o.DiscountAmount), 0)
    , 2)                                                   AS Gross_Margin_Pct

FROM orders o
JOIN products p ON o.ProductID = p.ProductID
GROUP BY p.Category
ORDER BY Gross_Margin_Pct DESC
"""
run_query(q11)
```

> 💡 **Insight:** Sports has the highest gross margin at 39.55% and Beauty is close behind at 39.34% — these two categories are the most profitable per rupee sold. Books has the lowest margin at 34.61% despite having ₹1.34 Cr in net revenue, because the COGS for Books (₹88.12L) is really high. Basically Books is generating a lot of revenue but keeping very little of it. I'd avoid heavy discounting on Books since the margins are already thin. Sports and Beauty on the other hand can handle some promotional discounts without going into the red.

## 🚚 Query 12 — Order Status Distribution & Shipping Cost Analysis

**Business Question:** What percentage of orders are delivered, pending, or cancelled?

A high percentage of pending or cancelled orders signals **operational issues** — poor logistics, payment failures, or stock-outs. Monitoring shipping costs helps identify if **logistics expenses are eating into margins** and whether courier negotiations are needed.

```python
q12 = """
SELECT
    OrderStatus,
    COUNT(OrderID)                                           AS Total_Orders,
    ROUND(100.0 * COUNT(OrderID) / (SELECT COUNT(*) FROM orders), 2) AS Share_Pct,
    ROUND(SUM(TotalAmount), 2)                               AS Total_Revenue,
    ROUND(AVG(ShippingCost), 2)                              AS Avg_Shipping_Cost,
    ROUND(SUM(ShippingCost), 2)                              AS Total_Shipping_Cost
FROM orders
GROUP BY OrderStatus
ORDER BY Total_Orders DESC
"""
run_query(q12)
```

> 💡 **Insight:** About 69.75% of orders are delivered successfully — that's decent but there's room to improve. What's worrying is the cancellation rate at 10.65% (213 orders),That's ₹98L in lost revenue from cancellations alone. Cancelled orders also have a slightly higher shipping cost (₹80.19 vs ₹76.21 for delivered orders), which means some cancellations are happening after the item has already been shipped — double the cost, zero the revenue.

## 🎯 Query 13 — Discount Impact: High vs Low Discount Orders

**Business Question:** Are heavy discounts actually driving more revenue, or are they hurting profitability?

This query buckets orders by discount level and compares average order value and profit. If **high-discount orders generate lower total revenue**, it indicates that discounts are not being used strategically and the business could be leaking margin unnecessarily.

```python
q13 = """
SELECT
    CASE
        WHEN DiscountRate = 0           THEN '0% - No Discount'
        WHEN DiscountRate <= 0.10       THEN '1-10% - Low'
        WHEN DiscountRate <= 0.20       THEN '11-20% - Medium'
        WHEN DiscountRate <= 0.30       THEN '21-30% - High'
        ELSE                                 '31%+ - Very High'
    END AS Discount_Bucket,
    COUNT(OrderID)                AS Total_Orders,
    ROUND(AVG(TotalAmount), 2)    AS Avg_Order_Value,
    ROUND(SUM(TotalAmount), 2)    AS Total_Revenue,
    ROUND(SUM(DiscountAmount), 2) AS Total_Discount_Given
FROM orders
GROUP BY Discount_Bucket
ORDER BY Total_Revenue DESC
"""
run_query(q13)
```

> 💡 **Insight:** This was a really interesting finding — customers who bought at full price (no discount) spent the most on average: ₹52,509 per order. Once discounts kick in, the average order value actually drops. The 21-30% discount bucket has the lowest AOV at ₹41,208. And those 359 orders in the high discount bucket received ₹36.92L in discounts. So the discounts are not making people spend more — they're just giving away money. The takeaway is clear: stop blanket discounting. Use deep discounts only to recover abandoned carts, not on customers who would have bought anyway.

## 📉 Query 14 — Products with Highest Return Rate

**Business Question:** Which specific products are being returned the most?

Products with a high return rate signal **quality issues, misleading descriptions, or sizing problems**. This analysis enables the business to take direct action — improving product pages, contacting suppliers, or temporarily pulling a product from listings until the issue is resolved.

```python
q14 = """
SELECT
    o.ProductID,
    p.Category,
    p.SubCategory,
    COUNT(DISTINCT o.OrderID)          AS Total_Orders,
    COUNT(DISTINCT r.ReturnID)         AS Total_Returns,
    ROUND(
        100.0 * COUNT(DISTINCT r.ReturnID) / NULLIF(COUNT(DISTINCT o.OrderID), 0)
    , 2)                               AS Return_Rate_Pct,
    ROUND(SUM(r.RefundAmount), 2)      AS Total_Refunded
FROM orders o
JOIN products p ON o.ProductID = p.ProductID
LEFT JOIN returns r ON o.OrderID = r.OrderID
GROUP BY o.ProductID, p.Category, p.SubCategory
HAVING COUNT(DISTINCT r.ReturnID) > 0
ORDER BY Return_Rate_Pct DESC
LIMIT 10
"""
run_query(q14)
```

> 💡 **Insight:** The highest return rate (27.27%) belongs to Home & Kitchen Pillows — 3 returns out of just 11 orders, with ₹1.03L refunded. Electronics Headphones are second at 26.67%. What really caught my eye is BTY-SUN-003 (Sunscreen) appearing in BOTH the top-rated products list AND this high return list at 25% return rate. It has a 5.0 rating but still gets returned 1 in 4 times — that screams product-description mismatch. People expect something different from what they receive.

## 🔁 Query 15 — Repeat Customer Analysis (Single vs Multi-Order Customers)

**Business Question:** How many customers have placed more than one order? What is their revenue contribution?

Repeat customers are the backbone of sustainable e-commerce. **Acquiring a new customer costs 5–7x more than retaining an existing one.** This query shows what proportion of revenue comes from loyal repeat buyers vs one-time buyers, directly informing **retention and loyalty program strategy**.

```python
q15 = """
WITH customer_orders AS (
    SELECT
        CustomerID,
        COUNT(OrderID)             AS Order_Count,
        ROUND(SUM(TotalAmount), 2) AS Total_Spend
    FROM orders
    GROUP BY CustomerID
)
SELECT
    CASE
        WHEN Order_Count = 1                    THEN 'One-Time Buyer'
        WHEN Order_Count BETWEEN 2 AND 4        THEN 'Occasional Buyer (2-4 orders)'
        ELSE                                         'Loyal Buyer (5+ orders)'
    END                            AS Customer_Type,
    COUNT(CustomerID)              AS Num_Customers,
    ROUND(SUM(Total_Spend), 2)     AS Total_Revenue,
    ROUND(AVG(Total_Spend), 2)     AS Avg_Spend_Per_Customer,
    ROUND(AVG(Order_Count), 2)     AS Avg_Orders_Per_Customer
FROM customer_orders
GROUP BY Customer_Type
ORDER BY Total_Revenue DESC
"""
run_query(q15)
```

> 💡 **Insight:** The numbers here are pretty eye-opening. Loyal Buyers — just 189 customers — are responsible for ₹4.98 Cr which is 55% of total revenue. Each of them spends an average of ₹2.64L. Meanwhile the 33 one-time buyers only contribute ₹1.38 Cr. The real opportunity is the 272 Occasional Buyers sitting in the middle — they spend ₹1.44L on average but haven't become loyal yet. If even 20% of them could be converted to loyal buyers, that's potentially ₹2+ Cr in additional revenue. A simple email follow-up sequence at Day 7, 30, and 60 after a purchase would be the cheapest and smartest way to make that happen.
